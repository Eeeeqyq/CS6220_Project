```python
from pathlib import Path
import pandas as pd

# Repo root = first parent folder containing README.md (works from the root or from notebooks/)
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "README.md").exists())
RAW = ROOT / "data" / "raw"

df = pd.read_csv(RAW / "2024_01.csv", dtype={"CRS_DEP_TIME": str}, low_memory=False)

# the four airports; use ORIGIN if the download has it, otherwise map from ORIGIN_AIRPORT_ID
airports = {10721: "BOS", 11618: "EWR", 12478: "JFK", 14100: "PHL"}
if "ORIGIN" in df.columns:
    df = df[df["ORIGIN"].isin(airports.values())].copy()
else:
    df = df[df["ORIGIN_AIRPORT_ID"].isin(airports)].copy()
    df["ORIGIN"] = df["ORIGIN_AIRPORT_ID"].map(airports)
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"], format="%m/%d/%Y %I:%M:%S %p").dt.date

print("Flights from the 4 airports in Jan 2024:", len(df))
print(df["ORIGIN"].value_counts())

cols = ["FL_DATE", "OP_UNIQUE_CARRIER", "ORIGIN", "DEST", "CRS_DEP_TIME",
        "DISTANCE", "DEP_DELAY", "ARR_DELAY", "ARR_DEL15", "CANCELLED"]

# one random flight per airport + one late flight, so the table shows both labels
sample = pd.concat([
    df.groupby("ORIGIN").sample(1, random_state=1),
    df[df["ARR_DEL15"] == 1].sample(1, random_state=1),
])[cols]
print(sample.to_string(index=False))
```
